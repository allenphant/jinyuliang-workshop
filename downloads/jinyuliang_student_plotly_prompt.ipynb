## 使用方式

1. 先準備 `jinyuliang_student_release_1.zip`；人體、食品檢體與環境資料會在後續幕次才釋出。
2. 依調查順序閱讀每一幕的前因、目前資料與公衛問題。
3. 將該幕 Prompt 完整貼給 Colab 裡的 Gemini。
4. Gemini 會先說明資料、分母與分析方向，再提供一格可執行的 Python 程式。
5. 執行後先閱讀表格與圖，再使用檢查 Prompt 核對病例、分母、時間與原始計數。
6. 將自己的公衛判讀寫在每幕最後的「學員判讀」區。

### AI 協作約定

- 學員提出調查問題、病例定義、分母與判讀需求。
- AI 先查看檔名、欄位說明與資料內容，再選擇程式寫法與內部名稱。
- 每一幕完成後，AI 要列出它建立了哪些分析成果，讓下一幕能沿用。
- 未知資料保留為未知；所有比例同時呈現分子與分母。
- 程式報錯時，將完整錯誤訊息與實際欄位貼回 Gemini，請它根據現況修正。

你不需要預先知道變數或函式名稱。你的核心任務是確認 AI 使用了正確的資料、病例定義與分母，並判斷結果支持到哪一層。
